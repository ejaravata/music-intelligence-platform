```python
album_mask = df_grammy["award"].str.contains(r"album", case=False, na=False)

song_mask = (
    df_grammy["award"].str.contains(r"song|single|record", case=False, na=False)
    & ~album_mask
)

artist_mask = (
    df_grammy["award"].str.contains(r"artist", case=False, na=False)
    & ~album_mask
    & ~song_mask
)
```

```python
df_albums = df_grammy.loc[album_mask, [
    "name", "year", "award", "artist", "annual_edition", "winner"
]]
df_songs = df_grammy.loc[song_mask, [
    "name", "year", "award", "artist", "annual_edition", "winner"
]]
df_artists = df_grammy.loc[artist_mask, [
    "year", "award", "name", "winner"
]]

df_artists = df_artists.dropna(subset=["name"])
```

```python
df_songs = df_songs.rename(columns={
    "name": "song_title",
    "artist": "artist_name"
})

df_albums = df_albums.rename(columns={
    "name": "album_title",
    "artist": "artist_name"
})

df_artists = df_artists.rename(columns={
    "name": "artist_name",
})
```

```python
#Replace all collaboration indicators with commas
df_billboard["artist"] = df_billboard["artist"].str.replace(
    r"(?i)\s*(?:&|feat\.?|ft\.?|featuring|x|\+)\s*",
    ",",
    regex=True
)

#Split into lists
df_billboard["artist"] = df_billboard["artist"].str.split(",")

#Expand into multiple rows
df_billboard = df_billboard.explode("artist")

#Clean whitespace
df_billboard["artist"] = df_billboard["artist"].str.strip()

#Rename artist to be artist_name
df_billboard = df_billboard.rename(columns={
    "artist": "artist_name",
    "date": "WeekEndingDate",
    "peak-rank": "PeakRank",
    "weeks-on-board": "WeeksOnBoard",
    "rank": "CurrentRank",
    "song": "SongName"
})

#merge!
df_billboard = df_billboard.merge(
    artists_info[["artist_name", "artist_id"]],
    left_on="artist_name",
    right_on="artist_name",
    how="left"
).drop(columns="artist_name")

#drop duplicates
df_billboard = df_billboard.drop_duplicates()

#drop any where we can't find the artist
df_billboard = df_billboard.dropna(subset=["artist_id"])

#drop id column
df_billboard = df_billboard.drop(columns=['id', 'last-week'])
```
